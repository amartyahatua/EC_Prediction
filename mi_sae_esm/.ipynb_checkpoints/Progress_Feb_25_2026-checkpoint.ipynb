### Data & Model Setup
* Loaded SwissProt-EC-leaf dataset (enzyme proteins with EC class labels)
* Filtered to proteins ≤512 residues, took 10K sequences

### Extract representations
* Ran each protein through ESM-2 8M model (6 layers, 320 dimensions)
* Extracted token-level representations from all layers
* Each residue in a protein becomes a 320-dimensional vector

### Layer selection — which layer has the most biological information?
* For each layer (0-6), mean-pooled representations across residues to get one vector per protein
* Trained logistic regression to predict EC class at all 4 hierarchy levels
* Layer 5 was the best across all levels (79.4% to 85.0%)
* Layer 0 was near random, Layer 6 dropped — Layer 5 is the sweet spot
* Conclusion: use Layer 5 representations for SAE training

### Hyperparameter for SAE
* Tested 12 configurations: 3 dict sizes (4×, 8×, 16×) × 4 TopK values (64, 128, 256, 512)
* All trained for 100 epochs, Adam optimizer (lr=1e-4), batch size 256
* Finding: TopK value controls reconstruction quality far more than dict size
* k=256 gives good reconstruction (MSE=0.005) while keeping features sparse
* k=512 reconstructs perfectly but is too dense to interpret
* Selected: dict=16× (5120 features), k=256

### Train final SAE
* Input: Layer 5 token-level representations (320-dim per residue)
* Output: 5120 features per residue, only 256 active at a time
* Auxiliary loss (coeff=0.1) to prevent dead features
* Learning rate halved if loss plateaus for 5 epochs
* Final reconstruction MSE = 0.005
* Saved as sae_dict16_k256.pt

### Feature analysis (20 proteins)
* For each protein, passed residue representations through trained SAE
* Got activation matrix: (seq_len × 5120) — how strongly each feature fires at each position
* Took top 10 most active features per protein

### EC specificity analysis (100 proteins, 5 EC classes)
* Picked 5 EC classes: *'EC:2.7.7.6', 'EC:4.2.1.11', 'EC:6.1.1.17', 'EC:3.6.4.12', 'EC:2.7.11.1'*
* 20 proteins per class
* For each protein, computed mean SAE feature vector (average across all residues → one 5120-dim vector)
* For each feature, computed F-ratio.
* Top features have F-ratio up to 28.3 — they strongly distinguish enzyme classes.

### Linear probe on SAE features
* Used mean SAE feature vectors (5120-dim) to predict EC class
* Trained logistic regression, 70/30 train/test split
* Result: 96.7% accuracy (compared to 82.7% on raw ESM-2 representations)

### Progressive causal ablation
* Question: how few features do you actually need to predict EC class?
* For each EC class, ranked features by how important they are (from linear probe weights)
* Kept only top N features, zeroed everything else, measured if prediction still works

## Future Steps

### Train SAEs on all 6 layers
* Train the same SAE (dict=16×, k=256) on layers 1, 2, 3, 4, 5, and 6
* Then run the linear probe on each layer's SAE features
* This shows whether deeper layers learn more functional features — same idea as the layer selection plot but with SAE features instead of raw representations



### Fair comparison: raw representations vs SAE features
* Pick the same *k*-EC classes, same proteins, same train/test split
* Run logistic regression on raw Layer *n* representations (320-dim)
* Run logistic regression on SAE features (5120-dim)
* Now we can compare: are SAE features better, worse, or similar?
